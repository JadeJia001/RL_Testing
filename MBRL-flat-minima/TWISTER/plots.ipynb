```python
import matplotlib.pyplot as plt
import numpy as np

# Data provided
rho = [0.0001, 0.00025, 0.0005, 0.001, 0.0025, 0.005, 0.01, 0.025, 0.05]
scores = [2110, 2103, 2100, 1905, 2820, 2528, 2175, 2399, 1730]

# Create the plot
plt.figure(figsize=(10, 6))

# Plot data with markers and a line
plt.plot(rho, scores, marker='o', linestyle='-')

# Set the x-axis to a logarithmic scale
plt.xscale('log')

# Add titles and labels for clarity
plt.title('Scores vs. Rho')
plt.xlabel('Rho (log scale)')
plt.ylabel('Scores')

# Add a grid for better readability of the log scale
plt.grid(True, which="both", ls="--")

# Display the plot
plt.show()
```
